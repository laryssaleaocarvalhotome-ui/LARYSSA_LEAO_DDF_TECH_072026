# ETAPA 10 – Indicadores de Negócio (KPIs)

## Objetivo

Construir os principais indicadores utilizados pela área de Operações para acompanhar o desempenho do e-commerce.

Nesta etapa serão calculados:

- Número de clientes
- Número de pedidos
- Número de produtos
- Número de vendedores
- Quantidade de categorias
- Faturamento total
- Ticket médio
- Nota média das avaliações

```python
# KPI 1 - Total de clientes únicos

total_clientes = clientes["customer_unique_id"].nunique()

print("=" * 50)
print("KPI 1 - Clientes Únicos")
print("=" * 50)
print(f"Total de clientes únicos: {total_clientes:,}")
```

```python
# KPI 2 - Total de pedidos

total_pedidos = pedidos["order_id"].nunique()

print("=" * 50)
print("KPI 2 - Total de Pedidos")
print("=" * 50)
print(f"Total de pedidos: {total_pedidos:,}")
```

```python
# KPI 3 - Total de produtos

total_produtos = produtos["product_id"].nunique()

print("=" * 50)
print("KPI 3 - Total de Produtos")
print("=" * 50)
print(f"Total de produtos: {total_produtos:,}")
```

```python
# KPI 4 - Total de vendedores

total_vendedores = vendedores["seller_id"].nunique()

print("=" * 50)
print("KPI 4 - Total de Vendedores")
print("=" * 50)
print(f"Total de vendedores: {total_vendedores:,}")
```

```python
# KPI 5 - Total de categorias

total_categorias = categorias["product_category_name"].nunique()

print("=" * 50)
print("KPI 5 - Total de Categorias")
print("=" * 50)
print(f"Total de categorias: {total_categorias:,}")
```

```python
# KPI 6 - Faturamento total

faturamento_total = itens["price"].sum()

print("=" * 50)
print("KPI 6 - Faturamento Total")
print("=" * 50)
print(f"Faturamento total: R$ {faturamento_total:,.2f}")
```

```python
# KPI 7 - Valor total do frete

frete_total = itens["freight_value"].sum()

print("=" * 50)
print("KPI 7 - Valor Total do Frete")
print("=" * 50)
print(f"Valor total do frete: R$ {frete_total:,.2f}")
```

```python
# KPI 8 - Ticket médio

ticket_medio = faturamento_total / total_pedidos

print("=" * 50)
print("KPI 8 - Ticket Médio")
print("=" * 50)
print(f"Ticket médio: R$ {ticket_medio:,.2f}")
```

```python
# KPI 9 - Nota média das avaliações

nota_media = avaliacoes["review_score"].mean()

print("=" * 50)
print("KPI 9 - Nota Média das Avaliações")
print("=" * 50)
print(f"Nota média das avaliações: {nota_media:.2f}")
```

```python
# Resumo dos KPIs

kpis = pd.DataFrame({
    "Indicador": [
        "Clientes únicos",
        "Pedidos",
        "Produtos",
        "Vendedores",
        "Categorias",
        "Faturamento Total",
        "Frete Total",
        "Ticket Médio",
        "Nota Média"
    ],
    "Valor": [
        f"{total_clientes:,}",
        f"{total_pedidos:,}",
        f"{total_produtos:,}",
        f"{total_vendedores:,}",
        f"{total_categorias:,}",
        f"R$ {faturamento_total:,.2f}",
        f"R$ {frete_total:,.2f}",
        f"R$ {ticket_medio:,.2f}",
        f"{nota_media:.2f}"
    ]
})

display(kpis)
```

## 11.1 Distribuição de Clientes por Estado

Objetivo:

Identificar os estados com maior quantidade de clientes cadastrados.

```python
import matplotlib.pyplot as plt

clientes_estado = (
    clientes.groupby("customer_state")
             .size()
             .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

clientes_estado.plot(kind="bar")

plt.title("Quantidade de Clientes por Estado")

plt.xlabel("Estado")

plt.ylabel("Quantidade de Clientes")

plt.xticks(rotation=45)

plt.show()
```

### Insight

Observa-se que o estado de São Paulo concentra a maior parte dos clientes da plataforma, seguido por Rio de Janeiro e Minas Gerais.

Esse comportamento é esperado devido à maior densidade populacional e ao elevado número de consumidores nessas regiões.

Esse tipo de informação pode auxiliar a área de Operações na priorização logística, planejamento de estoque e definição de estratégias comerciais regionais.

## 11.2 Categorias Mais Vendidas

### Objetivo

Identificar as categorias de produtos com maior volume de vendas.

```python
# Juntando os itens do pedido com os produtos
categorias_vendidas = itens.merge(
    produtos,
    on="product_id",
    how="left"
)

# Contando quantos produtos foram vendidos por categoria
top_categorias = (
    categorias_vendidas
    .groupby("product_category_name")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

# Criando o gráfico
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

top_categorias.plot(kind="bar", color="steelblue")

plt.title("Top 10 Categorias Mais Vendidas")

plt.xlabel("Categoria")

plt.ylabel("Quantidade Vendida")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

### Insight

As categorias apresentadas concentram a maior parte das vendas realizadas na plataforma.

Essas categorias representam oportunidades para ações estratégicas, como reforço de estoque, campanhas promocionais e negociações com fornecedores.

Também podem ser utilizadas como referência para planejamento logístico e definição de prioridades operacionais.

```python
plt.figure(figsize=(12,6))

top_categorias.sort_values().plot(
    kind="barh",
    color="#2E86C1"
)

plt.title("Top 10 Categorias Mais Vendidas", fontsize=16)

plt.xlabel("Quantidade Vendida")

plt.ylabel("Categoria")

plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()

plt.show()
```

### Insight

As categorias **Cama, Mesa e Banho**, **Beleza e Saúde** e **Esporte e Lazer** concentram o maior volume de vendas da plataforma.

Esse comportamento sugere que esses segmentos possuem elevada demanda e devem receber atenção especial das áreas de Operações e Logística, principalmente no planejamento de estoque, abastecimento e distribuição.

Além disso, essas categorias representam oportunidades para campanhas promocionais e estratégias de fidelização, uma vez que concentram uma parcela significativa das vendas.

## 11.3 Evolução Mensal dos Pedidos

### Objetivo

Analisar como a quantidade de pedidos evoluiu ao longo do período da base de dados, identificando tendências, crescimento ou sazonalidade.

```python
# Converter a coluna de data para o formato de data
pedidos["order_purchase_timestamp"] = pd.to_datetime(
    pedidos["order_purchase_timestamp"]
)

# Criar coluna Ano-Mês
pedidos["ano_mes"] = pedidos["order_purchase_timestamp"].dt.to_period("M")

# Contar pedidos por mês
pedidos_mes = (
    pedidos
    .groupby("ano_mes")
    .size()
)

# Transformar índice em texto
pedidos_mes.index = pedidos_mes.index.astype(str)

# Criar gráfico
plt.figure(figsize=(14,6))

pedidos_mes.plot(
    kind="line",
    marker="o",
    linewidth=2,
    color="#2E86C1"
)

plt.title("Evolução Mensal dos Pedidos", fontsize=16)

plt.xlabel("Ano/Mês")

plt.ylabel("Quantidade de Pedidos")

plt.grid(True, linestyle="--", alpha=0.4)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

### Insight

A análise da evolução mensal dos pedidos permite identificar tendências de crescimento, estabilidade ou queda na demanda ao longo do tempo.

Essas informações podem apoiar o planejamento operacional, o dimensionamento de equipes, o controle de estoque e a preparação para períodos de maior demanda.

```python
# Converter a data
pedidos["order_purchase_timestamp"] = pd.to_datetime(
    pedidos["order_purchase_timestamp"]
)

# Criar coluna Ano-Mês
pedidos["ano_mes"] = pedidos["order_purchase_timestamp"].dt.to_period("M")

# Contar pedidos por mês
pedidos_mes = (
    pedidos.groupby("ano_mes")
           .size()
           .reset_index(name="quantidade")
)

# Transformar em string
pedidos_mes["ano_mes"] = pedidos_mes["ano_mes"].astype(str)

# Remover meses com poucos registros (meses incompletos)
pedidos_mes = pedidos_mes[pedidos_mes["quantidade"] > 100]

# Criar gráfico
plt.figure(figsize=(14,6))

plt.plot(
    pedidos_mes["ano_mes"],
    pedidos_mes["quantidade"],
    marker="o",
    linewidth=2,
    color="#2E86C1"
)

plt.title("Evolução Mensal dos Pedidos", fontsize=16)

plt.xlabel("Ano/Mês")

plt.ylabel("Quantidade de Pedidos")

plt.grid(True, linestyle="--", alpha=0.4)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

### Insight

Observa-se um crescimento gradual da quantidade de pedidos ao longo do período analisado, indicando expansão da operação da plataforma.

Após um período inicial de crescimento, o volume mensal de pedidos se estabiliza em um patamar elevado, sugerindo maturidade da operação.

Os meses finais da base apresentam menor quantidade de registros por representarem períodos incompletos e, por isso, não foram considerados na análise de tendência.

# ETAPA 12 – Tempo de Entrega

## Objetivo

Avaliar a eficiência logística da operação por meio do cálculo do tempo entre a compra e a entrega do pedido.

Serão analisados:

- Tempo médio de entrega
- Distribuição dos tempos
- Possíveis atrasos
- Insight operacional

```python
# Converter datas para formato datetime

pedidos["order_purchase_timestamp"] = pd.to_datetime(
    pedidos["order_purchase_timestamp"]
)

pedidos["order_delivered_customer_date"] = pd.to_datetime(
    pedidos["order_delivered_customer_date"]
)

# Criar coluna com dias de entrega

pedidos["tempo_entrega"] = (
    pedidos["order_delivered_customer_date"] -
    pedidos["order_purchase_timestamp"]
).dt.days

pedidos.head()
```

```python
tempo_entrega = pedidos["tempo_entrega"].dropna()

print("Quantidade de pedidos entregues:", len(tempo_entrega))
```

```python
media_entrega = tempo_entrega.mean()

print("=" * 50)
print("TEMPO MÉDIO DE ENTREGA")
print("=" * 50)

print(f"Média: {media_entrega:.2f} dias")
```

```python
print("=" * 50)
print("ESTATÍSTICAS DO TEMPO DE ENTREGA")
print("=" * 50)

print(f"Mínimo: {tempo_entrega.min()} dias")
print(f"Máximo: {tempo_entrega.max()} dias")
print(f"Média: {tempo_entrega.mean():.2f} dias")
print(f"Mediana: {tempo_entrega.median():.2f} dias")
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    tempo_entrega,
    bins=30,
    edgecolor="black"
)

plt.title("Distribuição do Tempo de Entrega")
plt.xlabel("Dias")
plt.ylabel("Quantidade de Pedidos")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

### Insight

Observa-se que a maior parte dos pedidos foi entregue em um intervalo relativamente curto, concentrando-se entre aproximadamente 5 e 20 dias.

Poucos pedidos apresentaram tempos de entrega significativamente superiores à média, formando uma cauda à direita na distribuição. Esses casos podem estar relacionados a fatores como distância geográfica, indisponibilidade de estoque, problemas logísticos ou atrasos de transportadoras.

No geral, os resultados indicam uma operação logística eficiente, com boa previsibilidade no prazo de entrega para a maior parte dos clientes. O acompanhamento contínuo desse indicador permite identificar gargalos operacionais e direcionar ações para reduzir atrasos e melhorar a experiência do consumidor.
